```python
limit = 100
pokemon = []

for page_num in range(14):
    print("----")
    offset = page_num * limit
    url = f"https://pokeapi.co/api/v2/pokemon?offset={offset}&limit={limit}"  
    print("Requesting", url)

    response = requests.get(url)
    data = response.json()
    pokemon.extend(data['results'])
```

```python
url = "https://pokeapi.co/api/v2/pokemon?limit=100"

while url:
    print("Requesting", url)
    response = requests.get(url)
    data = response.json()
    url = data['next']
```

```python
# Method: Listing to the API
# Pros: The API knowns going on inside of it, good if you don't know how many pages there are
# Cons: The API has to actually tell you the next page
```
